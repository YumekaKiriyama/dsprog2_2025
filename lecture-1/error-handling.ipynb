エラーハンドリングの基礎

```python
x = 1
y = 0
```

```python
# 0で割っているのでエラーが起きる
x / y
```

```python
try:
    resulut = x / y
except ZeroDivisionError as e:
    resulut = None
    print("エラーが発生しました：", e)
    print("0で割ることはできません。結果をNoneに設定します。")
    
    print("結果：",resulut)
```

### 複数のエラーが発生する可能性がある場合

```python
try:
    resulut = x / z
except ZeroDivisionError as e:
    resulut = None
    print("エラーが発生しました：", e)
    print("0で割ることはできません。結果をNoneに設定します。")
except NameError as e:
    resulut = None
    print("エラーが発生しました：", e)
    print("未定義の変数が使用されました。結果をNoneに設定します。") 

    print("結果：",resulut)
```

### エラーが発生しなかった場合にのみ実行したい処理がある場合

```python
z = 3

try:
    resulut = x / z
except ZeroDivisionError as e:
    resulut = None
    print("エラーが発生しました：", e)
    print("0で割ることはできません。結果をNoneに設定します。")
except NameError as e:
    resulut = None
    print("エラーが発生しました：", e)
    print("未定義の変数が使用されました。結果をNoneに設定します。") 
else:
    # tryブロック内でエラーが発生しなかった場合に実行される
    print("エラーは発生しませんでした。計算結果を表示しました。")

    print("結果：",resulut)
```

### エラーの発生の有無に関わらず実行したい処理がある場合

```python
z = 3

try:
    resulut = x / z
except ZeroDivisionError as e:
    resulut = None
    print("エラーが発生しました：", e)
    print("0で割ることはできません。結果をNoneに設定します。")
except NameError as e:
    resulut = None
    print("エラーが発生しました：", e)
    print("未定義の変数が使用されました。結果をNoneに設定します。") 
else:
    # tryブロック内でエラーが発生しなかった場合に実行される
    print("エラーは発生しませんでした。計算結果を表示しました。")
finally:
    # エラーの発生の有無に関わらず実行される
    print("計算処理が終了しました。")

    print("結果：",resulut)
```

### システム終了以外の全てのエラーをハンドリングしたい場合

```python
x = 1
y = 0   
z = 3

try:
    resulut = x / y
except Exception as e:
    resulut = None
    print("エラーが発生しました：", e)
    print("エラーオブジェクト", type(e))
    print("エラーが発生したため、結果をNoneに設定します。")
else:
    print("エラーは発生しませんでした。計算結果を表示しました。")
finally:
    print("計算処理が終了しました。")

    print("結果：",resulut)
```

### Webスクレイピングでのエラー処理

```python
import requests
```

```python
url = 'https://www.musashino-u.ac.jp/invalid-page'
response = requests. get (url)
print (f"ステータスコード：{response.status_coder}")
```

```python
url = 'https://www.musashino-u.ac.jp/invalid-page' # 存在しないページURL

try:
    response = requests.get(url)
    response.raise_for_status()  
except requests.exceptions.HTTPError as e:
    print(f"HTTPエラーが発生しました：", e)
except requests.exceptions.RequestException as e:
    print(f"リクエストエラーが発生しました：", e)   
else:
    print("Webページに正常にアクセスできました。")
finally:
    print("ウェブスクレイピング処理が終了しました。")
```
