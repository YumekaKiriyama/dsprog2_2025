### DB（SQLite）を使う時の処理の流れ

1. DBへのコネクションを確立
2. コネクションオブジェクトからデータ操作用のカーソルオブジェクトを取得
3. 実行したいSQLを定義
4. SQLを実行
5. 必要があればコミットする
6. コネクションを閉じる

```python
path = ''
db_name = 'car.db'

try:
    conn = sqlite3.connect(path + db_name)
    
    cur = conn.cursor()
  # SQL(RDBを操作するための言語)の実行するためのカーソルオブジェクトを取得

    # SQLの定義
    # テーブルの作成
    sql = 'CREATE TABLE cars (id INT, name TEXT, price REAL);'

    # SQL文の実行
    cur.execute(sql)

except sqlite3.Error as e:
    print(f'エラーが発生しました:',e)

finally:    
    # DBへの変更を閉じる
    conn.close()
```

### データをDBに挿入する（単一レコード）

```python
path = ''
db_name = 'car.db'

try:
    conn = sqlite3.connect(path + db_name)
    
    cur = conn.cursor()
  # SQL(RDBを操作するための言語)の実行するためのカーソルオブジェクトを取得

    # SQLの定義
    # テーブルの作成
    sql = "INSERT INTO cars (id, name, price) VALUES (1, 370, 7000000);"

    # SQL文の実行
    cur.execute(sql)

except sqlite3.Error as e:
    print(f'エラーが発生しました:',e)

finally:    
    # DBへの変更を閉じる
    conn.close()
```

### データ

```python
path = ''
db_name = 'car.db'

try:
    conn = sqlite3.connect(path + db_name)
    cur = conn.cursor()

    # SQL（レコード挿入）
    sql = "INSERT INTO cars (id, name, price) VALUES (?, ?, ?);"

    cars = [
        (2, 'GT-R', 15000000),
        (3, 'シビックタイプR', 5000000),
        (4, 'シビック', 4000000)
    ]

    # 複数レコードを挿入する場合は executemany を使う
    cur.executemany(sql, cars)

    conn.commit()

except sqlite3.Error as e:
    print('エラーが発生しました:', e)

finally:
    conn.close()
```

### DB内のデータを参照する

```python
path = ''
db_name = 'car.db'

try:
    conn = sqlite3.connect(path + db_name)
    cur = conn.cursor()

    # SQL（レコード挿入）
    sql = "SELECT * FROM cars;"

    # 複数レコードを挿入する場合は executemany を使う
    cur.execute(sql)

    #conn.commit()

except sqlite3.Error as e:
    print('エラーが発生しました:', e)

else:
    for row in cur:
        id, name, price = row
        print(id, name, price)

finally:
    conn.close()
```

```python
path = ''
db_name = 'car.db'

try:
    conn = sqlite3.connect(path + db_name)
    cur = conn.cursor()

    # SQL（レコード挿入）
    # DELETE FROM テーブル名 WHERE id = 1;
    sql = "DELETE FROM cars WHERE id = 1;"

    # 複数レコードを挿入する場合は executemany を使う
    cur.execute(sql)

    conn.commit()

except sqlite3.Error as e:
    print('エラーが発生しました:', e)

finally:
    conn.close()
```

```python
path = ''
db_name = 'car.db'

try:
    conn = sqlite3.connect(path + db_name)
    cur = conn.cursor()

    # SQL（レコード挿入）
    
    sql = "UPDATE cars SET price = ? WHERE id = ?;"

    # 複数レコードを挿入する場合は executemany を使う
    cur.execute(sql, (5500000, 2))

    conn.commit()

except sqlite3.Error as e:
    print('エラーが発生しました:', e)

finally:
    conn.close()
```
